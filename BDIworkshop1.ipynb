# **DATABASES**


What are DBs?

A **Database (DB)** is an organized collection of data that can be easily accessed, managed, and updated.  
Think of it like a digital filing cabinet that stores information in a structured way, making it easier to retrieve and use.

**Example**:  
A library database stores book records, including title, author, genre, and availability.

---

**Why are they needed?**
1. **Organized Data Storage**: Keeps data structured and easy to access.  
   *Example*: Customer details in an e-commerce platform.
2. **Efficient Data Retrieval**: Quickly fetch specific information.  
   *Example*: Searching for all orders placed in the last month.
3. **Data Integrity**: Ensures data is accurate and consistent.  
   *Example*: Preventing duplicate entries for the same customer.
4. **Multi-User Access**: Supports multiple users accessing the database simultaneously.  
   *Example*: Employees in a company accessing the same HR database.
5. **Scalability**: Handles growing amounts of data effectively.  
   *Example*: A social media platform managing millions of user profiles.

---


**Types of Databases**
1. **Relational Databases (RDBMS)**:
   - Data is stored in **tables** with rows and columns.
   - Uses SQL (Structured Query Language) for data manipulation.
   - **Best for structured data** like customer information, inventory, etc.
   - **Examples**: MySQL, PostgreSQL, Oracle Database, SQL Server.

   **Use Case**:  
   E-commerce platforms for managing user accounts, orders, and payments.

2. **NoSQL Databases**:
   - Non-tabular; stores data in key-value pairs, documents, or graphs.
   - Flexible schema, good for unstructured or semi-structured data.
   - **Examples**: MongoDB, Cassandra, DynamoDB, Redis.

   **Use Case**:  
   Social media platforms for storing posts, comments, and user interactions.

3. **Graph Databases**:
   - Data is stored as nodes and relationships (edges).
   - Ideal for analyzing connections and relationships.
   - **Examples**: Neo4j, Amazon Neptune.

   **Use Case**:  
   Recommendation systems like "People You May Know" on LinkedIn.

4. **Time-Series Databases**:
   - Optimized for storing time-stamped data.
   - **Examples**: InfluxDB, TimescaleDB.

   **Use Case**:  
   IoT sensor data, stock prices, and server logs.

5. **Object-Oriented Databases**:
   - Data is stored as objects, like in object-oriented programming.
   - **Examples**: Db4o, ObjectDB.

   **Use Case**:  
   Applications with complex data relationships like CAD systems.

---

**When to Use What Database?**

| **Type**              | **Use Case**                                                                                                                                     | **Example DBs**                |
|------------------------|--------------------------------------------------------------------------------------------------------------------------------------------------|---------------------------------|
| Relational DBs         | Structured data, financial systems, CRM, and ERP systems                                                                                       | MySQL, PostgreSQL, SQL Server  |
| NoSQL DBs             | High scalability, unstructured data, real-time analytics                                                                                       | MongoDB, DynamoDB              |
| Graph DBs             | Social networks, fraud detection, recommendation engines                                                                                       | Neo4j, Amazon Neptune          |
| Time-Series DBs       | Monitoring systems, IoT devices, performance tracking                                                                                         | InfluxDB, TimescaleDB          |
| Object-Oriented DBs   | Applications requiring object-oriented data storage (e.g., 3D modeling or multimedia)                                                          | ObjectDB, Db4o                 |

Types:

*   Inner
*   Left

*   Outer
*   Natural
*   Cross


**-- Formula for outer joins:**

*Left join* = inner join + fetch any additional records from left table which is not present in right table.

*Right join* = inner join + fetches any records from right table which is not present in left table.

*Full join* = inner join
                + fetch additional records from left table which is not present in right table
                + fetch additional records from right table which is not present in left table.

# **Python**

**Python** is a high level object oriented language used for machine learning, analytics, web development and many other use cases.

**What is OOPs?**

Object-Oriented Programming (OOP) is a way of writing code where you organize it around objects rather than actions or logic. Objects are like real-world things that have attributes (data) and behaviors (functions or methods). For example, think of a car:

***Object: Car***

*Attributes (data)*: Color, brand, speed, fuel level

*Behaviors (methods)*: Drive, brake, refuel

OOP is built around 4 key principles:

*   Encapsulation
*   Inheritance
*   Abstraction
*   Polymorphism



**Encapsulation**: Keeping all the details (data and methods) related to an object in one place. For example, all details about a "Car" are grouped together.

**Inheritance**: Creating new objects by building on existing ones. For example, a "SportsCar" can inherit features from the general "Car" object but also add its own special features like "turbo mode."

**Polymorphism**: Making objects behave differently based on their type. For instance, a "Drive" method for a "Car" and a "Bike" might work differently but still share the same name.

**Abstraction**: Hiding unnecessary details and showing only what's important. For example, when you drive a car, you don’t need to know how the engine works; you just use the steering wheel and pedals.

In short, OOP makes it easier to design and organize complex programs by thinking in terms of "things" and what they do, just like in the real world.

## Python Data Structures

# Python Lists

- **Lists** are used to store multiple items in a single variable.
- They are:
  - Ordered: The order of items is preserved.
  - Mutable: You can change the items after creation.
  - Allow duplicates: You can have the same value multiple times.
- **Syntax**: `[item1, item2, item3]`

#### Example Use Cases:
- Storing a collection of similar items (e.g., a list of fruits).
- Keeping track of tasks or to-do lists.


````python
# Creating and using a list
fruits = ["apple", "banana", "cherry"]
print("Original list:", fruits)

# Modifying a list
fruits.append("orange")  # Add item
fruits[1] = "blueberry"  # Modify item
print("Modified list:", fruits)

#To access a list item

````

# Tuples and Sets

Python Tuples
- **Tuples** are used to store multiple items, like lists, but they are **immutable** (unchangeable).
- They are:
  - Ordered: The order of items is preserved.
  - Allow duplicates: You can have the same value multiple times.
- **Syntax**: `(item1, item2, item3)`

Example Use Cases:
- Storing items that should not change, like coordinates or configurations.
- Using tuples as dictionary keys (since they are immutable).


````python
# Creating and using a tuple
colors = ("red", "green", "blue")
print("Original tuple:", colors)

# Accessing tuple elements
print("First color:", colors[0])

# Tuples are immutable
# colors[0] = "yellow"  # Uncommenting this line will throw an error
````

Python Sets
- **Sets** are used to store unique items in an unordered way.
- They are:
  - Unordered: Items have no defined order.
  - Mutable: You can add or remove items.
  - No duplicates allowed: Adding duplicate items has no effect.
- **Syntax**: `{item1, item2, item3}`

Example Use Cases:
- Removing duplicates from a list.
- Performing set operations like union, intersection, and difference.


````python
# Creating and using a set
numbers = {1, 2, 3, 4}
print("Original set:", numbers)

# Adding and removing elements
numbers.add(5)
numbers.remove(3)
print("Modified set:", numbers)

# Sets do not allow duplicates
numbers.add(2)  # Adding duplicate does nothing
print("After adding duplicate:", numbers)
````

````python
# You can also convert lists to tuples
l = [1,2,2,3,4,5]
s = set(l)
s
````

# Python Dictionaries



- **Dictionaries** store data in **key-value pairs**.
- They are:
  - Mutable: You can change or add key-value pairs.
  - Unordered: The order is maintained only in Python 3.6+.
  - Keys must be unique, but values can repeat.
- **Syntax**: `{key1: value1, key2: value2}`

Example Use Cases:
- Storing attributes of an object (e.g., a student's name, age, and grade).
- Creating lookup tables or mappings.

````python
# Creating and using a dictionary
student = {"name": "John", "age": 20, "grade": "A"}
print("Original dictionary:", student)

# Accessing and modifying values
print("Student name:", student["name"])
student["age"] = 21  # Update value
student["city"] = "New York"  # Add new key-value pair
print("Modified dictionary:", student)
````

# Modules

What are Python Modules?
A module in Python is a file that contains Python code, such as functions, variables, and classes. Modules are used to organize and reuse code across multiple programs.

You can:

Use built-in modules like math and random.
Create your own custom modules.
Import modules into your program using the import keyword.


Why Use Modules?

Modules help in:


*   Keeping your code organized.
*   Using libraries created by others.
*   Avoiding code duplication.






````python
# Import the math module
import math

# Use a function from the math module
result = math.sqrt(16)  # Square root of 16
print("The square root of 16 is:", result)
````

````python
# Import only the sqrt function from the math module
from math import sqrt

# Use the imported function
result = sqrt(25)
print("The square root of 25 is:", result)
````

# Control Flow


Control Flow refers to the order in which the code is executed in a program. Python provides several control flow statements to make decisions, execute code conditionally, or repeat actions.

The three main types of control flow statements are:

* Conditional Statements (if, elif, else)
* Loops (for, while)
* Control Statements (break, continue, pass)

Conditional Statements

Conditional statements allow you to execute code only when certain conditions are met.

**Loops**
Python supports two types of loops:

* **for loop**: Used to iterate over a sequence (like a list, tuple, or string).
* **while loop**: Repeats code as long as a condition is True.


**Control Statements**

Control statements modify the behavior of loops:

* break: Exits the loop entirely.
* continue: Skips the current iteration and moves to the next.
* pass: Does nothing; a placeholder.
